```python
df = pd.DataFrame({
    'Rollno.':[1,2,3,4,5,6,7,3,8,9,10],
    'Name':['Shardul','Ram','Sita','Lakshaman','Arjun','Bheem','Nakul','Sita','Karn','Balram','Krishna'],
    'Class':['A',None,'C','B','C','A','C','C',None,'B','A'],
    'Age':[16,16,17,16,15,16,17,17,16,15,16],
    'Marks':[56,88,78,37,77,None,59,78,90,83,79]
})
```

```python
# .loc :     row index and column name
df.loc[3,'Marks']
```

```python
# change values
# change marks of Krishna

df.iloc[10,4] = 95
```

```python
df
```

```python
# show students whose marks are above 85

df[df['Marks'] > 85]
```

```python
# show students whose marks are above 75 and are in class B

df[(df['Marks'] > 75) & (df['Class'] == 'B')]
```

```python
# find in Class B how much students age in 16

df[(df['Age'] == 16) & (df['Class'] == 'B')]
```

```python
# use OR (|)
# find whre is null basis on class and  marks

df[(df['Class'].isnull()) | (df['Marks'].isnull())]
```

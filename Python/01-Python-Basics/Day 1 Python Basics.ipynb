```python
z = 'Python Programming'
# find a and ' ' (by positive indexing)
# find n and h   (by negative indexing)

print(z[12])
print(z[6])
print()
print(z[-2])
print(z[-15])
```

```python
# Positive Slicing [min:max+1]
z = 'Python Programming'
print(z[0:6])
print(z[7:18])
print(z[7:])
print(z[:])
print(z[:14])
```

```python
# Negative Slicing
print(z[-11:])
print(z[-11:-4])
```

```python
# stepping
z = 'Python Programming'

print(z[3:14:2])
print(z[3:14:3])
```

```python
# prog to find out whether a string consist of vowels or not

text = input("Enter a string: ").lower()
# if 'a' or 'e' or 'i' or 'o' or 'u' in text:
if (
    ("a" in text)
    or ("e" in text)
    or ("i" in text)
    or ("o" in text)
    or ("u" in text)
):
    print("contains vowels.")
else:
    print("not contain vowels.")
```

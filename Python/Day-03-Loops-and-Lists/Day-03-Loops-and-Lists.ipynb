```python
# grede system:
# 91 to 100 : O
# 81 to 90: A
# 71 to 80 : B
# 61 to 70: C
# < 60 : Fail

marks = int(input("Enter your Marks:"))

if marks >= 91 and marks <= 100:
    print('You got O Grade')
elif 81 <= marks and marks <= 90:
    print('You got A Grade')
elif 71 <= marks and marks <= 80:
    print('You got B Grade')
elif 61 <= marks and marks <= 70:
    print('You got C Grade')
elif marks <= 60:
    print('You got F Grade')
else:
    print('Enter valid Marks')
```

```python
# program to find even and odd number from a list

l = [7,8,2,9,10,15,17]
evens = [num for num in l if num % 2 == 0]
odds = [num for num in l if num % 2 != 0]

print("Even:", evens)
print("Odd:", odds)
```

```python
l = [7, 8, 2, 9, 10, 15, 17]
evens=[]
odds=[]
for num in l:
    if num % 2 == 0:
        evens.append(num)
    else:
        odds.append(num)

print("Even:", evens)
print("Odd:", odds)
```

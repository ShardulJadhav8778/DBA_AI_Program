```python
# program to find even and odd number from a list taking list as a input from a user

l = input("Enter the list using (,):")
l = l.split(',')
for i in l:
    if int(i)%2==0:
        print("Even:",i)
    else:
        print("Odd:",i)
```

```python
# menue system taking multiple order and printing total bill after order is complete

print("MENUE")
print("1.Mango \-20rs\n2.Apple\-30rs\n3.Strabery\-25rs\n4.Banana\-15rs\n5.Berry\-40rs")

total = 0

while True:
    order = int(input("Enter the Order number(0 to exit):"))
    if order == 0:
        break
    quantity = int(input("Enter the Quantity:"))
    if order == 1:
        total = total + (20*quantity)
        print('Mango is ordered')
    elif order == 2:
        total = total + (30*quantity)
        print('Apple is ordered')
    elif order == 3:
        total = total + (25*quantity)
        print('Strawbery is ordered')
    elif order == 4:
        total = total + (15*quantity)
        print('Banana is ordered')
    elif order == 5:
        total = total + (40*quantity)
        print('Berry is ordered')
    else:
        print('Invalid order')

print("Total Bill:",total)
```

```python
# create a menu system taking 
# multiple orders and printing 
# total bill after the order is
# completed

print('-------MENU------')
print('1. Vadapav')
print('2. Samosa')
print('3. Tea')
print('4. Poha')
cont = 'y'
total = []
q = {}
price = {'vadapav': 15,'samosa':20,'tea': 12, 'poha' : 25 }

while True:
  
  order = input('Enter The Order Number (Exit:[n]) : ')

  if order == 'n':
    break

  if int(order) == 1:  # false
    qty = int(input('Enter The Quantity : '))
    total.append(15*qty)
    q['vadapav'] = qty
  elif int(order) == 2:  # False
    qty = int(input('Enter The Quantity : '))
    total.append(20*qty)
    q['samosa'] = qty
  elif int(order) == 3:  # False
    qty = int(input('Enter The Quantity : '))
    total.append(12*qty)
    q['tea'] = qty
  elif int(order) == 4:   # False
    qty = int(input('Enter The Quantity : '))
    total.append(25*qty)
    q['poha'] = qty
  else:
    print('Order does not exist')
    total.append(0)

print('---- Total Bill----') 
for i,j in q.items():
  print(f'{i} x {j} = {price[i] * j}')
print('The Total Bill is  = ', sum(total))
```

```python
# Write a program to guess a number
# example
# range from 1,10
# num = 7
# guessed number = 3 ----low
# guessed number = 9 ----high
# five chances to guess correct number

import random
num = random.randint(1,10)
print(num)

for i in range(5):
    guess = int(input("Guess number from 1 to 10:"))
    if guess == num:
        print("Correct Guess")
        break
    elif guess < num:
        print(f'guessed number = {guess} --- low')
        if num - guess <= 2:
            print(f'guessed number = {guess} --- Too close')
        else:
            print(f'guessed number = {guess} --- Too far')
    else:
        print(f'guessed number = {guess} --- high')
        if guess - num <= 2:
            print(f'guessed number = {guess} --- Too close')
        else:
            print(f'guessed number = {guess} --- Too far')

else:
    print("Your all 5 chances completed")
```

```python
# program to reverse a given number using a loop

num = int(input("Enter a number: "))
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reversed number:", rev)
```

```python
# program to check whether the given number is palindrome number or not

num = int(input("Enter a number: "))
original = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if original == rev:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

```python
# program to check whether the given number is Armstrong number or not

num = int(input("Enter a number: "))
original = num
n = len(str(num))
sumn = 0

while num > 0:
    digit = num % 10
    sumn = sumn + digit ** n
    num = num // 10

if original == sumn:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

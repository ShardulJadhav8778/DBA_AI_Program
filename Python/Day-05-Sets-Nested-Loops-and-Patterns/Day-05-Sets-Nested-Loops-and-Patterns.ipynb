```python
# Write a program to guess a number
# example
# range from 1,10
# num = 7
# guessed number = 3 ----low
# guessed number = 9 ----high
# five chances to guess correct number

import random
num = random.randint(1,10)

for i in range(5):
    guess = int(input("Guess number from 1 to 10:"))
    if guess == num:
        print("Correct Guess")
        break
    elif guess < num:
        print(f'guessed number = {guess} --- low')
        if num - guess <= 2:
            print(f'guessed number = {guess} --- Too close')
        else:
            print(f'guessed number = {guess} --- Too far')
    else:
        print(f'guessed number = {guess} --- high')
        if guess - num <= 2:
            print(f'guessed number = {guess} --- Too close')
        else:
            print(f'guessed number = {guess} --- Too far')

else:
    print("Your all 5 chances completed")
    print(f'Correct Number was {num}')
```

```python
# program to reverse a list without using functions & slicing

num = [1,2,3,4,5]
rev = []

i = len(num) - 1

while i >= 0:
    rev.append(num[i])
    i = i - 1

print("Reverse:",rev)
```

```python
# program to print a dictionary with help of while loop

j = {'One':'Yellow', 'Two':'Red', 'Three':'Blue'}

keys = list(j.keys())
i = 0

while i < len(keys):
    print(keys[i], j[keys[i]])
    i = i + 1
```

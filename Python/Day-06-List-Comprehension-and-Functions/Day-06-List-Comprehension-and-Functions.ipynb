```python
# create user defined function max

def large(l):
    maxn = l[0]
    for i in l:
        if i > maxn:
            maxn = i
    return maxn

num = input("Enter the number with space:")
l = [int(i) for i in num.split()]
print(large(l))
    
```

```python
# program to crate a user defined function just like inbuilt len function

def length(l):
    count = 0
    for i in l:
        count = count+1
    return count
l = input("Enter:")
# l = [int(i) for i in l.split()]
print(length(l))
```

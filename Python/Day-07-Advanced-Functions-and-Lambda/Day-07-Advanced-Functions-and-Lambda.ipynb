```python
# write a function to check wherther the number is ambstrong number or not

def armstrong(n):
    original = n
    n = len(str(n))
    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit ** n
        n = n // 10

    if original == total:
        print('Armstrong number')
    else:
        print('Not Armstrong number')

n = int(input('Enter number:'))
armstrong(n)
```

```python
# write a function to reverse a number without using string reversing

def reverse(n):
    rev = 0

    while n > 0:
        digit = n % 10
        rev = rev*10 + digit
        n = n // 10

    return rev

n = int(input('Enter number:'))
reverse(n)
```

```python
# Arbitary Argument Function *  (data stored in Tuple)
# Keyword Argument Function **  (data stored in dictionary)

def length(*args):
    print(args)
    print(type(args))

length(5,4,3,2,1,6,8)
```

```python
# create a Arbitary Argument Function to find a sum of list

def addition(*args):
    total = 0

    for i in args:
        total = total + i

    return total

print(addition(1,2,3,4,5))
```

```python
# Keyword Argument Function **  (data stored in dictionary)

def student_info(**dic):
    print(dic)
    print(type(dic))

student_info(Rollno = 1,Name = 'Shardul',Class = 'SYMCA')
```

```python
def student_info(**dic):
    for i,j in dic.items():
        print(f'{i} - {j}')

student_info(Rollno = 1,Name = 'Shardul',Class = 'SYMCA')
```

```python
# Lambda Function
# Lambda Function also called Anonymous Function
# it is defined using lambda keyword
# its is a one line funcation
# lambda functions by default run return
```

```python
x = lambda a : a**2
print(x(7))
```

```python
# write a lambda function to add Two numbers

a = lambda x,y : x + y
print(a(2,4))
```

```python
print((lambda x,y : x + y)(3,2))
```

```python
v = [2,4,6]
t = lambda x : [i**2 for i in x]
print(t(v))
```

```python
# Exception Handling 
```

```python
for i in range(5,-5,-1):
    try:
        print(10/i)
    except:
        print('Logical Error')
```

```python

try:
    print(int('Hello'))
except:
    print('Check the code')
```

```python
try:
    print(int('Hello'))
except ValueError:
    print('Check the code')
```

```python
try:
    print(15/0)
except ValueError:
    print('Check the code')
```

```python
try:
    print(15/0)
except ZeroDivisionError:
    print('Divide by zero')
```

```python
try:
    print(15/0)
except Exception:  # Father of all exvetion, used as bydefault
    print('Divide by zero')
```

```python
# finally: find try block(1,2,3,....)

try:
    print(15/0)
except:
    print('Divide by zero')
finally: 
    print('Hello')
```

```python
try:
    print(15/5)
except:
    print('Divide by zero')
finally:
    print('Try Block 1')
```

```python
class Car:               # creating a class
    c_name = 'Tata'      # Class attribute

    def __init__(self,m_name):
        self.m_name = m_name    # object attribute

c1 = Car('Nexon')
print(c1.c_name)
print(c1.m_name)
print()

c2 = Car('Namo')
print(c2.c_name)
print(c2.m_name)
```
